In this exercise, we will work with the `spam.csv` data file. This file contains text and a label attached to each text. The goal is to clean the text data and then build classifcation models. 

### Exercise 1(a) (2 points)

Load the `pandas` and `numpy` libraries.

```python
import pandas as pd
import numpy as np
```

### Exercise 1(b) (2 points)

Read the `csv` file and create a data-frame called `spam_df`.

```python
spam_df = pd.read_csv("spam.csv")
spam_df.head()
```

### Exercise 1(c) (2 points)

Using the `string` library, remove all the punctuation of the `text` column.

```python
import string

spam_df['text_clean'] = spam_df['text'].str.replace(f"[{string.punctuation}]", "", regex=True)
spam_df.head()
```

### Exercise 1(d) (10 points)

Using the `nltk` library, remove stopwords and pronouns from `text_clean` and store in another column called `'text_clean_1'`. 

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_words = set(stopwords.words('english'))
pronouns = ['i', 'me', 'my', 'mine', 'you', 'your', 'yours', 'he', 'him', 'his', 'she', 'her', 'hers', 'we', 'us', 'our', 'ours', 'they', 'them', 'their', 'theirs', 'u', 'ur', 'im', 'dont', '2', 'ill', '4']

def clean_text(text):
    #tokenize the text
    tokens = word_tokenize(text.lower())

    #remove unwanted tokens (stop words and pronouns)
    tokens = [token for token in tokens if token not in stop_words and token not in pronouns]

    #replacing words
    tokens = [word.replace('txt', 'text') for word in tokens]

    return " ".join(tokens)

spam_df['text_clean_1'] = spam_df['text_clean'].apply(clean_text)
spam_df.head()
```

### Exercise 1(e) (10 points)

Perform Lemmatization of `text_clean_1` and store the results in another column called `text_clean_2` in the `spam_df` data frame.

```python
from nltk.corpus import wordnet
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

#download necessary resources
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')

#initialize lemmatizer
lemmatizer = WordNetLemmatizer()

#function to get the part of speech tag for lemmatization
def get_wordnet_pos(word):
    tag = nltk.pos_tag([word])[0][1][0].upper()
    tag_dict = {"J": wordnet.ADJ,
                "N": wordnet.NOUN,
                "V": wordnet.VERB,
                "R": wordnet.ADV}
    return tag_dict.get(tag, wordnet.NOUN)

#function to lemmatize text
def lemmatize_text(text):
    tokens = word_tokenize(text)
    lemmatized_words = [lemmatizer.lemmatize(token, get_wordnet_pos(token)) for token in tokens]
    return " ".join(lemmatized_words)

#apply the function to lemmatize the text
spam_df['text_clean_2'] = spam_df['text_clean_1'].apply(lemmatize_text)
spam_df.head()
```

### Exercise 1(f) (5 points)

Based on the results from part 1(f), we see digits in some of the text in `text_clean_2`. Remove all the digits from `text_clean_2` and store the results in another column `text_clean_3` in the `spam_df` data frame.

```python
import re

#function to remove numbers from text
def remove_numbers(text):
    return re.sub(r'\d+', '', text)

#apply the function to 'text_clean_2' to remove numbers from the text
spam_df['text_clean_3'] = spam_df['text_clean_2'].apply(remove_numbers)
spam_df.head()
```

### Exercise 1(g) (5 points)

Generate topic features using BERT embeddings on `text_clean_3`.

```python
import matplotlib.pyplot as plt
import seaborn as sns

from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

print("Loading BERT model...")
bert_model = SentenceTransformer('all-MiniLM-L6-v2')
print("BERT model loaded.")

#generating the embeddings for the cleaned text
print("Generating embeddings...")
embeddings = bert_model.encode(spam_df['text_clean_3'].tolist(),
                               show_progress_bar=True,
                               batch_size=32)
print("Embeddings generated.")
print(f"Embeddings shape:", embeddings.shape)
print(f"Each document is represented by a {embeddings.shape[1]}-dimensional vector")
```

### Exercise 1(h) (5 points)

Apply the K-Means clustering algorithm to the bert embedding generated in the previous step.

1. Set the number of clusters to k=3.

2. Assign the cluster labels to a new column in `spam_df` called `'kmeans_cluster_label'`.

3. Generate new features by computing the distance of each document to the centroid of each of the k=3 clusters. Store these distances in a NumPy array called `X_kmeans_features`.

```python
from sklearn.preprocessing import normalize

#normalizing the embeddings
embeddings_normalized = normalize(embeddings)

#apply KMeans clustering on the normalized embeddings
num_clusters = 3
kmeans = KMeans(n_clusters=num_clusters, n_init=20, random_state=42)
kmeans.fit(embeddings_normalized)

#storing the cluster labels
spam_df['cluster'] = kmeans.labels_

#compute the distance of each point to its assigned cluster centerr
X_kmeans_features = kmeans.transform(embeddings_normalized)
```

### Exercise 1(i) (5 points)

Hierarchical clustering is a type of density-based grouping used to organize documents based on similarity.

1. Apply Agglomerative Hierarchical Clustering to the L2​-normalized BERT embeddings (bert_embeddings) with k=3 clusters, using the `'ward'` linkage method.

2. Store the resulting cluster assignments in a new column called `'bert_hierarchical_label'` in `spam_df`.

3. Generate new features by computing the distance of each document to the centroid of each of the k=3 clusters. Store these distances in a NumPy array called `X_hierarchical_features`.

```python
from sklearn.cluster import AgglomerativeClustering

#apply agglomerative clustering on the normalized embeddings
num_clusters = 3
agg_clustering = AgglomerativeClustering(n_clusters=num_clusters, linkage="ward")
spam_df['bert_historical_label'] = agg_clustering.fit_predict(embeddings_normalized)
print(spam_df.head())
```

```python
from sklearn.metrics.pairwise import euclidean_distances

#compute the centroids manually
centroids = []
for i in range(num_clusters):
    centroids.append(embeddings_normalized[spam_df['cluster'] == i].mean(axis=0))
centroids = np.array(centroids)

#compute the distances from each point to the centroids
X_hierarchical_features = euclidean_distances(embeddings_normalized, centroids)
```

### Exercise 1(j) (10 points)

Train and evaluate six `LogisticRegression` models as follows:

- Using `kmeans` features as input.
- Using `hierarchical` features as input.
- Using bert embedding features as input.
- Combining `kmeans` + bert embeddings features as input.
- Combining `hierarchical` + bert embeddings features as input.
- Combining `kmeans` + `hierarchical` + bert embeddings features as input.

Evaluate the six models over a five-fold cross validation strategy.

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

y = spam_df['label'].map({'spam': 1, 'ham': 0})
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

#first input
X_kmeans = X_kmeans_features

#second input
X_hierarchical = X_hierarchical_features

#third input
X_third = embeddings

#fourth input
X_fourth = np.hstack([X_kmeans, embeddings])

#fifth input
X_fifth = np.hstack([X_hierarchical, embeddings])

#sixth input
X_sixth = np.hstack([X_kmeans, X_hierarchical, embeddings])

print("Comparing classification performance of different input combinations")
print("="*60)

model = make_pipeline(StandardScaler(), LogisticRegression(random_state=42))

#evaluate each feature set
feature_sets = {
    "First input": X_kmeans,
    "Second input": X_hierarchical,
    "Third input": X_third,
    "Fourth input": X_fourth,
    "Fifth input": X_fifth,
    "Sixth input": X_sixth
}

results = {}
for name, X in feature_sets.items():
    scores = cross_val_score(model, X, y, cv=skf, scoring='accuracy', n_jobs=-1)
    results[name] = (scores.mean(), scores.std())
    mean, std = results[name]
    print(f"{name}: Mean accuracy = {mean:.4f}, STD = {std:.4f}")

#finding the best performing feature set
best_feature_set = max(results, key=lambda k: results[k][0])
print(f"Best performing feature set: {best_feature_set} with Mean accuracy = {results[best_feature_set][0]:.4f}, STD = {results[best_feature_set][1]:.4f}   ")
```
